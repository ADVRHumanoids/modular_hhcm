```python
from modular.URDF_writer import UrdfWriter
```

```python
urdf_writer = UrdfWriter(speedup=True)
homing_joint_map = {}
```

```python
urdf_writer.add_module('socket.yaml')

# J1
data = urdf_writer.add_module('module_joint_yaw_ORANGE.yaml')
homing_joint_map[data['name']] = 0.0

# J2
data = urdf_writer.add_module('module_joint_double_elbow_ORANGE.yaml')
homing_joint_map[data['name']] = 0.5

# J3
data = urdf_writer.add_module('module_joint_yaw_ORANGE.yaml')
homing_joint_map[data['name']] = 0.0

# J4
data = urdf_writer.add_module('module_joint_double_elbow_ORANGE.yaml')
homing_joint_map[data['name']] = 1.0

# J5
data = urdf_writer.add_module('module_joint_yaw_ORANGE.yaml')
homing_joint_map[data['name']] = 0.0

# J6
data = urdf_writer.add_module('module_joint_double_elbow_ORANGE.yaml')
homing_joint_map[data['name']] = 1.5

# Simple end effector
urdf_writer.add_simple_ee(z_offset=0.189)
```

```python
urdf_writer.remove_all_connectors()
urdf_writer.write_urdf()
urdf_writer.write_lowlevel_config()
urdf_writer.write_problem_description_multi()
# Use the same collision sampling budget as the generator CLI.
urdf_writer.write_srdf(builder_joint_map=homing_joint_map, num_trials=1000)
urdf_writer.write_joint_map()
```

```python
urdf_writer.deploy_robot("modularbot6")
```
